```python
import pandas as pd
sprzedaż_09_10 = pd.read_csv('online_retail_09_10.csv', sep=',')
sprzedaż_10_11 = pd.read_csv('online_retail_10_11.csv', sep=',')
print(sprzedaż_09_10.columns.values)
print(sprzedaż_10_11.columns.values)
print(all(sprzedaż_09_10.columns == sprzedaż_10_11.columns))
sprzedaż = pd.concat([sprzedaż_09_10, sprzedaż_10_11], ignore_index=True)
```

```python
print(sprzedaż_09_10.shape) #wynik: (525461, 8)
print(sprzedaż_10_11.shape) #wynik: (541910, 8)
print(sprzedaż.shape) #wynik: (1067371, 8)
print(sprzedaż_09_10.shape[0] + sprzedaż_10_11.shape[0] == sprzedaż.shape[0]) #wynik: true
```

```python
zwroty = sprzedaż['InvoiceNo'].str.startswith('C')
francja = sprzedaż['Country'] == 'France'

print(sprzedaż.loc[zwroty & francja, ['InvoiceNo','CustomerID']])
klient_12615 = (sprzedaż['CustomerID'] == 12615.0)

print(sprzedaż.loc[zwroty & klient_12615, ['Description', 'StockCode']])
print(sprzedaż[klient_12615 & (sprzedaż['StockCode'] == '22618')])
```

```python
zwroty = sprzedaż['InvoiceNo'].str.startswith('C')
print(sum(zwroty)) #wynik: 19494
print(sum(zwroty) / len(sprzedaż)) #wynik: 0.018
print(sprzedaż.loc[zwroty, "InvoiceNo"].nunique()) #wynik: 8292

print((sprzedaż.loc[zwroty, 'Quantity'] < 0).all()) #wynik: False
print(sprzedaż.loc[zwroty & (sprzedaż["Quantity"] > 0)])
```
